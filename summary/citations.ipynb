# 12. Precision, Recall, F1-score, and Balanced Accuracy

**Pedregosa, F., et al. (2011).**  
*Scikit-learn: Machine Learning in Python.*  
Journal of Machine Learning Research, **12**, 2825–2830.

**scikit-learn documentation:** *Metrics and scoring: quantifying the quality of predictions.*

**Used for:** Calculation and interpretation of classification precision, recall, F1-score, and balanced accuracy.

**Methodological role:**  
- **Precision** measures how often predicted members of a class are correct.  
- **Recall** measures how many true members of a class are recovered.  
- **F1-score** is the harmonic mean of precision and recall.  
- **Balanced accuracy** is the average recall across classes and is useful when class counts are unequal.

# 14. Random Forest Feature Importance — Mean Decrease in Impurity

**Breiman, L. (2001).**  
*Random Forests.*  
Machine Learning, **45**, 5–32.  
DOI: **10.1023/A:1010933404324**

**scikit-learn documentation:** *Feature importances with a forest of trees.*

**Used for:** The built-in `feature_importances_` values from `RandomForestClassifier`.

**Methodological role:** In scikit-learn, impurity-based feature importance is the normalized total reduction in the tree-splitting criterion attributable to each feature, accumulated across the trees in the forest. This is also commonly called **mean decrease in impurity (MDI)** or Gini importance when Gini impurity is used.

# 15. Permutation Feature Importance

**Breiman, L. (2001).**  
*Random Forests.*  
Machine Learning, **45**, 5–32.

**scikit-learn documentation:** *Permutation feature importance* and `sklearn.inspection.permutation_importance`.

**Used for:** Independent validation of Random Forest feature importance.

**Methodological role:** A feature is randomly shuffled while other columns remain unchanged. The resulting decrease in model performance estimates how much predictive information the fitted model obtains from that feature. This provides a useful comparison with impurity-based RF importance.

# 20. Eta / von Neumann Successive-Difference Statistic

**von Neumann, J., Kent, R. H., Bellinson, H. R., & Hart, B. I. (1941).**  
*The Mean Square Successive Difference.*  
The Annals of Mathematical Statistics, **12**(2), 153–162.  
DOI: **10.1214/aoms/1177731746**

**von Neumann, J. (1941).**  
*Distribution of the Ratio of the Mean Square Successive Difference to the Variance.*  
The Annals of Mathematical Statistics, **12**(4), 367–395.  
DOI: **10.1214/aoms/1177731677**

**Used for:** `EtaVonNeumann`, a time-order-sensitive variability feature.

**Methodological role:** The statistic compares squared differences between successive observations with the overall variance. Smooth/correlated light curves tend to have smaller successive differences than randomly ordered measurements, making η useful for characterizing temporal structure.
